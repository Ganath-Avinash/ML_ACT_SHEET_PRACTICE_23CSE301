```python
import pandas as pd

# Create the dataset
data = {
    "Brightness": [40, 50, 60, 10, 70, 60, 25],
    "Saturation": [20, 50, 90, 25, 70, 10, 80],
    "Class": ["Red", "Blue", "Blue", "Red", "Blue", "Red", "Blue"]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# Features and target
X = df[["Brightness", "Saturation"]]
y = df["Class"]

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

# Create KNN model
knn = KNeighborsClassifier(n_neighbors=3)

# Train the model
knn.fit(X_train, y_train)

# Predictions
y_pred = knn.predict(X_test)

print("Predictions:", y_pred)
print("Actual:", list(y_test))

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))

# Confusion Matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
new_sample = pd.DataFrame([[45, 30]], columns=['Brightness', 'Saturation'])   # Brightness=45, Saturation=30

prediction = knn.predict(new_sample)

print('Predicted Class:', prediction[0])
```
