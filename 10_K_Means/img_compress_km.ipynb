# K-Means Image Compression
**Dataset:** Sample 396x396 RGB image

## Importing Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from PIL import Image
```

## Loading and Resizing the Image

```python
img = Image.open('sample_image.jpg').convert('RGB')
img = img.resize((396, 396))
img = np.array(img)
print(img.shape)
```

## Displaying the Original Image

```python
plt.imshow(img)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

## Flattening Pixels for Clustering

```python
rows, cols, channels = img.shape
pixels = img.reshape(rows * cols, channels)
print(pixels.shape)
```

## Fitting K-Means (k=16 colors)

```python
k = 16
kmeans = KMeans(n_clusters=k, random_state=42, init='k-means++')
kmeans.fit(pixels)
```

## Reconstructing the Compressed Image

```python
compressed_pixels = kmeans.cluster_centers_[kmeans.labels_]
compressed_pixels = compressed_pixels.astype(np.uint8)
compressed_img = compressed_pixels.reshape(rows, cols, channels)
```

## Comparing Original vs Compressed

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Original Image')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(compressed_img)
plt.title('Compressed Image (K=' + str(k) + ')')
plt.axis('off')
plt.show()
```
